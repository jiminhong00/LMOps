```python
#open file in read mode

f_vanilla = open("./my_data/experiment/uprise/rte_0Shot_pred.json", "r")
f_uprise1 = open("./my_data/experiment/uprise/rte_Uprise_pred_1shot.json", "r")
f_uprise3 = open("./my_data/experiment/uprise/rte_Uprise_pred_3shot.json", "r")
f_uprise = open("./my_data/experiment/uprise/rte_Uprise_pred.json", "r")
```

```python
#parse json file
import json
vanilla = json.load(f_vanilla)
uprise1 = json.load(f_uprise1)
uprise3 = json.load(f_uprise3)
uprise = json.load(f_uprise)
```

```python
wrong_vanilla = 0
wrong_uprise = 0
right_vanilla = 0
right_uprise = 0
wrong_upper = 0
right_upper = 0
right_uprise1 = 0
right_uprise3 = 0

for (i, j) in zip(vanilla, uprise1):
    if i['label'] != i['pred'] and j['label'] != j['pred']:
        wrong_upper += 1
    else:
        right_upper += 1
        
for i in vanilla:
    if i['label'] != i['pred']:
        wrong_vanilla += 1
    else:
        right_vanilla += 1
for i in uprise:
    if i['label'] != i['pred']:
        wrong_uprise += 1
    else:
        right_uprise += 1
for i in uprise1:
    if i['label'] == i['pred']:
        right_uprise1 += 1
for i in uprise3:
    if i['label'] == i['pred']:
        right_uprise3 += 1
```

```python
print(right_uprise/len(uprise), right_vanilla/len(vanilla), right_upper/len(vanilla))
print(right_uprise1/len(uprise1), right_uprise3/len(uprise3))
```
